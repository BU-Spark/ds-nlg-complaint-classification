### This notebook is being used to merge the classification output of the llm with the cleaned full dataset

```python
import pandas as pd
import json
import csv
```

```python
import csv
import json

# Load JSON data from a file
with open('../initial testing/gpt-6-luna_testing.json', "r", encoding="utf-8") as json_file:
    data = json.load(json_file)

# Write to CSV
with open("../Joining Datasets/Data/classification_output.csv", "w", newline="", encoding="utf-8") as csv_file:
    # Extract column headers from the keys of the first dictionary
    fieldnames = data[0].keys()
    writer = csv.DictWriter(csv_file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(data)
    
```

```python
df_classified = pd.read_csv('../Joining Datasets/Data/classification_output.csv')
df_cleaned = pd.read_csv('../Joining Datasets/Data/all_complaints_collapsed.csv')
```

```python
df = df_cleaned.merge(df_classified, how= 'inner',left_on = 'ia_no', right_on = "id")
```
