### This notebook is being used to merge the classification output of the llm with the cleaned full dataset

```python
import pandas as pd
```

```python
df_classified = pd.read_csv('../LLM Classification/llm_results/results_llm_final_df.csv')
df_cleaned = pd.read_csv('../Joining Datasets/Data/all_complaints_collapsed.csv')
```

```python
df_classified.head()
```

```python
df_classified=df_classified.drop(columns=['run', 'prompt_version'])
```

```python
df = df_cleaned.merge(df_classified, how= 'inner',on = 'ia_no')
```

```python
df.head()
```

```python
df.to_csv('../LLM Classification/combined_classification_cleaned_df.csv')
```
