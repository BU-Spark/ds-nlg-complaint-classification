Using this document to clean and isolate 2019 data. 
this will later be merged with the 2011-2020 and 2020-2026 data. the 2011-2020 is missing 2019

```python
#importing libraries
import pandas as pd 
```

```python
df = pd.read_excel("Data/COMPLAINTS CONVERTED - 2016,17,19,20_corrected_file_final-Spark-spring2024 (1).xlsx")
df.head()
```

```python
selected = df[~df['IA No'].str.startswith('IAD2019', na=False)]
```

```python
df.drop(selected.index, inplace=True)
```

```python
df.head()
```

```python
#need to make sure that the description column is formated correctly so I can cleanly convert data into a csv file. 

df["Description"] = (df["Description"].str.replace(r"\s*[\r\n]+\s*", " ¶ ", regex=True))
```

```python
df = df.rename(columns={
    "IA No": "ia_no",
    "Incident type": "incident_type_x",
    "Received date": "received_date_x",
    "Description": "narrative_x",
    "Allegation": "allegation_x",
    "Finding": "finding_x",
    "First name": "Off: First name",
    "Last name": "Off: Last name",
    "Title/rank (snap)": "Off: Title/rank",
    "Action taken": "Act: Action taken"
})
df.head()
```

```python
df.to_csv("Data/cleaned_complaints_2019.csv", index=False)
```
